# Polynomial Regression

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt 
%matplotlib inline

print("Liabraries imported!!")
```

```python
# To use polynomial equation we needs to use quadratic equation : ax2 + bx + y = 0

x = 6 * np.random.rand(100, 1) - 3
y = 0.5 * x**2 + 1.5*x + 2 + np.random.randn(100,1) # quadratic equation 
#y = 0.5x^2 + 1.5x + 2 + outliers

plt.scatter(x,y)
plt.xlabel('X dataset')
plt.ylabel('y label')
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test = train_test_split(x,y, test_size = 0.2,random_state = 2)
```

```python
x_train
```

```python
## Implementing Simple Linear Regression

from sklearn.linear_model import LinearRegression
regression_1 = LinearRegression()
```

```python
regression_1.fit(x_train,y_train)
```

```python
from sklearn.metrics import r2_score
score = r2_score(y_test,regression_1.predict(x_test))
print(score) # Low Accuracy
```

```python
# Visualization

plt.plot(x_train, regression_1.predict(x_train), color = 'r')
plt.scatter(x_train,y_train)
plt.xlabel("X Dateset")
plt.ylabel("Y Dataset")
```

```python
## Applying Polynomial Transformation

from sklearn.preprocessing import PolynomialFeatures
```

```python
poly = PolynomialFeatures(degree=2, include_bias= True)
x_train_poly = poly.fit_transform(x_train) # Training X Data
x_test_poly = poly.transform(x_test) # Testing X Data
```

```python
x_train_poly
```

```python
x_test_poly
```

```python
from sklearn.metrics import r2_score
regression = LinearRegression()
regression.fit(x_train_poly, y_train)
y_pred = regression.predict(x_test_poly)
score = r2_score(y_test,y_pred)
print(score)
```

```python
plt.scatter(x_train,regression.predict(x_train_poly))
plt.scatter(x_train,y_train)
```

```python
poly = PolynomialFeatures(degree=3, include_bias= True)
x_train_poly = poly.fit_transform(x_train) # Training X Data
x_test_poly = poly.transform(x_test) # Testing X Data
```

```python
x_train_poly
```

```python
from sklearn.metrics import r2_score
regression = LinearRegression()
regression.fit(x_train_poly, y_train)
y_pred = regression.predict(x_test_poly)
score = r2_score(y_test,y_pred)
print(score)
```

```python
# Prediction of new data

x_new = np.linspace(-3,3,200).reshape(200,1)
x_new_poly = poly.transform(x_new)
y_new = regression.predict(x_new_poly)
plt.plot(x_new,y_new,'r-',linewidth = 2, label="New Predictions")
plt.plot(x_train,y_train,"b.",label = "Training Points")
plt.plot(x_test,y_test,"g.",label="Testing Points")
plt.xlabel("X")
plt.ylabel("Y")
plt.legend()
plt.show()
```

# Pipeline in Polynomial Regression

```python
from sklearn.pipeline import Pipeline
```

```python
### Here we'll combine two models together --> 1. PolyFeature and 2. LinearRegression model
def poly_regression(degree):
    x_new = np.linspace(-3,3,200).reshape(200,1)
    
    poly_features = PolynomialFeatures(degree = degree, include_bias=True)
    lin_reg = LinearRegression()
    poly_regression = Pipeline([
        ("poly_features", poly_features),
        ("lin_reg", lin_reg)
    ])
    
    poly_regression.fit(x_train, y_train)
    y_pred_new = poly_regression.predict(x_new)
    
    # Visualization
    
    plt.plot(x_new,y_pred_new,'r', label="Degree")
    plt.plot(x_train,y_train,'b.',linewidth=3)
    plt.plot(x_test, y_test,'g.',linewidth=3)
    plt.xlabel("X")
    plt.ylabel("Y")
    plt.axis([-4,4,0,10])
    
    plt.show()
```

```python
poly_regression(5)
```
