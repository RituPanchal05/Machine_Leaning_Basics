# House Price Prediction - Linear Regression

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn import linear_model

print("Liabraries imported sucessfully!")
```

```python
df = pd.read_csv("../data/house_price_prediction.csv")
print("Data file imported!!")
```

```python
plt.scatter(df.area, df.price, color='red', marker='+')

plt.xlabel("Area")
plt.ylabel("Price")


plt.show()
```

```python
reg = linear_model.LinearRegression()
reg.fit(df[['area']],df.price)
```

```python
predicted_price = reg.predict(pd.DataFrame([[3300]], columns=['area']))
```

```python
plt.figure(figsize=(8,5))

# Actual data
plt.scatter(df.area, df.price, color='red',marker='+',label='Actual Data')

# Regression line
plt.plot(df.area,reg.predict(df[['area']]),label='Regression Line')

# Predicted point
plt.scatter(3300,predicted_price,color='green',marker='*',label='Predicted Price')

plt.xlabel("Area")
plt.ylabel("Price")

plt.legend()
plt.show()
```
